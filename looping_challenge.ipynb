```python
#transactions = [1200, 450, 800, 1500, 2300, 700, 100]

#calculate total transaction value without using sum().

transactions = [1200, 450, 800, 1500, 2300, 700, 100]

temp =0
for i in transactions:
    temp = temp +i

print(temp)
```

```python
#From the same list, find the highest and lowest transaction without max() and min().

max = transactions[0]
min = transactions[0]
n = len(transactions)

for i in range(0,n):
    if transactions[i] > max:
        max = transactions[i]
    elif transactions[i] < min:
        min = transactions[i]

print(max)
print(min)
    



```

```python
#Given student marks

#marks = [78, 92, 45, 67, 88, 53, 99]

#count how many students scored 90+,75–89, 50–74, below 50

marks = [78, 92, 45, 67, 88, 53, 99]

student_A = 0
student_B = 0
student_c = 0
student_d = 0

for i in marks:
    if i >=90:
        student_A +=1
    elif i <=89 and i >=75:
        student_B +=1
    elif i <=74 and i >=50:
        student_c+=1
    else:
        student_d+=1

print("the number of students scored above 90 is",student_A)
print("the number of students scored between 75-89 is ",student_B)
print("the number of students scored between 50-74 is",student_c)
print("the number of students scored below 50 is ",student_d)

```

```python
#Count the frequency of every character in a string without using Counter.

#Example :banana ==> b -> 1 a -> 3 n -> 2

fruit_name = "banana"
n = len(fruit_name)
frequency = {}
for char in fruit_name:
    if char in frequency:
        frequency[char] = frequency.get(char,0)+1
    else:
        frequency[char] =1

print(frequency)
```

```python
#Find the second-largest number in a list without using sort().

num_list = [25,45,64,73,22,14]

max_num = float('-inf')
second_max = float('-inf')

for num in num_list:
    if num > max_num:
        second_max = max_num
        max_num = num

    elif num > second_max:
        second_max = num

print(second_max)
 

```

```python
#Check whether a string is a palindrome using loops.

s = "mad"
n = len(s)
reversed = []

for i in range(n-1, -1, -1):
    reversed.append(s[i])


reversed_string = "".join(reversed)
print(reversed_string)
print("the string is palindrome" if reversed_string == s else "this is not a palindrome")
```
