```python
student = {"name": "Rahul", "age": 22, "course": "Python", "city": "Bangalore", "marks": 88}
```

```python
#Print the complete dictionary
print(student)
```

```python
#Print the student's name
print(student['name'])
```

```python
#Print their course
print(student['course'])
```

```python
# Print all keys
print(student.keys())
```

```python
# Print all values
print(student.values())
```

```python
# Print all key-value pairs
print(student.items())
```

```python
# Change marks from 88 to 92
student['marks'] = 92
```

```python
print(student['marks'])
```

```python
# Add "email"
student['email'] = "ram@123"
```

```python
# Add "phone"
student["ph_num"] = "1234567"
```

```python
print(student)
```

```python
# Remove "city"

student.pop("city")
```

```python
print(student)
```

```python
# Use get() to retrieve "name"
student.get("name")
#student['name']
```

```python
#Create a copy of the dictionary
studnet_2 = student.copy()
```

```python
print(studnet_2)
```
