```python
# Create a function that accepts another function as an argument.

#Example idea calculate(add, 10, 20),calculate(multiply, 10, 20)

def add(a,b):
    return a+b

def multiply(a,b):
    return a*b

def calculate(a,b, fun):
    return fun(a,b)
```

```python
calculate(10,20, add)
```

```python
calculate(10,20, multiply)
```

```python
#Write a recursive function to calculate factorial.

def factorial(n):
    if n==0:
        return 1

    return n * factorial(n-1)
```

```python
factorial(5)
```

```python
# Write a recursive function to calculate 1 + 2 + 3 + ... +n

def recursive_sum(n):
    if n==0:
        return 0
    return n+ recursive_sum(n-1)
```

```python
recursive_sum(4)
```

```python
# Write a recursive function to generate the Fibonacci sequence or calculate the nth Fibonacci number.

def fibonacci_series(n):
    if n==0:
        return 0
    if n==1:
        return 1

    return  fibonacci_series(n-1) + fibonacci_series(n-2)
```

```python
fibonacci_series(4)   #(0,1,1,2,3)
```

```python
# Demonstrate local and global variable scope using a small program.

count = 0
def sum(l):
    global count
    temp =0

    for i in l:
        temp = temp+i
        count +=1
    return  temp

```

```python
print(sum([1,2,3,4,5]))
print(count)
```
