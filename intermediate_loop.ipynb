```python
#Print numbers from 1–100 but skip numbers divisible by 5 using continue.

n = 100

for i in range(1,n+1):
    if i%5 == 0:
        continue
    print(i)
```

```python
#Iterate from 1–100 and stop when you encounter the first number divisible by both 7 and 11.

n = 1
while n<=100:
    if n%7==0 and n%11==0:
        print("the first number which is divisible by both 7 and 11", n)
        break
    print(n)
    n = n+1
     
```

```python
# Search for a user-provided number inside a list. Use for-else to print:

#"Number Found","Number Not Found"

user_input = int(input("enter the number"))
li = [24,25,46,72,68,92,45]

for i in li:
    if i == user_input:
        print("number is found")
        break
else:
    print("number is not found")
```

```python
#Generate multiplication tables from 1 to 10 using nested loops.
n = 10
for i in range(1,n+1):
    for j in range(1, n+1):
        print(f"{i} * {j} = {i*j}")
```

```python
#Given a list containing duplicate elements, create another list containing only unique elements without using set().

nums = [24, 25, 43, 25, 63, 23, 24, 21]

n= len(nums)
original_list = []

for i in range(0,n):

    found = False
    m = len(original_list)
    for j in range(0, m):
        if nums[i] == original_list[j]:
            found = True
            break

    if found == False:
        original_list.append(nums[i])

print(original_list)
     


```

```python
#Given

#numbers = [10, -4, 8, -2, 0, 15, -9, 21]

#count ==> positive numbers, negative numbers, zeros

numbers = [10, -4, 8, -2, 0, 15, -9, 21]
positive_count = 0
negeative_count = 0
zeroes =0

for i in numbers:
    if i >0:
        positive_count += 1
    elif i<0:
        negeative_count +=1
    else:
       zeroes+=1

print(f"positive count : {positive_count}, negative_count : {negeative_count}, zeroes = {zeroes}")


```

```python
#Write a program to determine whether a number is prime using a loop.

number = int(input("enter a number"))

if number <= 1:
    print(f"{number} is not a prime number.")
else:
    is_prime = True
    
    for i in range(2, int(number ** 0.5) + 1): 
        if number % i == 0:
            is_prime = False 
            break              
            
    if is_prime:
        print(f"{number} is a prime number.")
    else:
        print(f"{number} is not a prime number.")

 
```
