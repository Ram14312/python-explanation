```python
python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}
```

```python
java_students = {"Priya", "Karan", "Rohit", "Simran"}
```

```python
# Print both sets

print(python_students)
```

```python
print(java_students)
```

```python
# Find students learning either Python or Java ( union())
python_students.union(java_students)
```

```python
# Find students learning both (#intersection)
python_students.intersection(java_students)
```

```python
# Find students learning only Python (#difference) (A-B = A)

python_students.difference(java_students)
```

```python
# Find students learning only Java (#difference) (B-A=B)
java_students.difference(python_students)
```

```python
# Find students belonging to exactly one group (symmetric difference) (a,b)

python_students.symmetric_difference(java_students)
```

```python
# Add a new student (add())
python_students.add("RAM")
```
