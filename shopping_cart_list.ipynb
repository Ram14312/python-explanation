```python
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]
```

```python
# Display all products
print(cart)
```

```python
# Access first and last products
print(cart[0])
print(cart[-1])
```

```python
# Add "Webcam"
cart.append("Webcam")
```

```python
print(cart)
```

```python
# Insert "USB Hub" at index 2
cart.insert(2,"USB Hub")
```

```python
print(cart)
```

```python
# Remove "Mouse"
cart.remove("Mouse")
```

```python
print(cart)
```

```python
# Remove the last item using pop()
cart.pop()
```

```python
print(cart)
```

```python
# Find the index of "Monitor"
cart.index("Monitor")
```

```python
# Count occurrences of "Laptop"
cart.count("Laptop")
```

```python
# Create a copy of the cart
cart_1 = cart
```

```python
print(cart_1)
print(cart)
```

```python
cart_1.remove("Monitor")
```

```python
print(cart_1)
print(cart)
```

```python
#shallow copy
cart_2 = cart.copy()
```

```python
print(cart_2)
print(cart)
```

```python
cart_2.remove("USB Hub")
```

```python
print(cart_2)
print(cart)
```

```python
# Reverse the cart
cart.reverse()
print(cart)
```

```python
# Sort the products alphabetically
cart.sort()
print(cart)

```

```python
#Explain the difference between
#append(), extend(), insert(), remove(), pop(), clear(), copy(), sort(), reverse()
#append() ==> so with the help of this method we can able to add a value to the list at the last index
# insert() ==> this method is used  for to add a new value into the list at specific index
# remove() ==> with the help of these remove method we can able to remove any specific element
# pop() ==> the pop funstion which is used to remove the element at the last index by default , if we want specific index we can able to do
# clear() ==> the clear() which is used to clear the all the values/elements from the list
#copy() ==> which is used to create a shallow copy of the original list
#sort() ==> which is used to sort the list in alphabetical order
# reverse() ==> we can able reverse the list
# extend method we can able to extend a iterable value to one by one index 
```
