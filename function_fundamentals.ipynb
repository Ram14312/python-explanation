```python
# Create a function that calculates factorial.

def factorial(n):
    if n==0:
        return 1
    return n * factorial(n-1)
```

```python
factorial(3)
```

```python
# Create a function that checks whether a number is prime.

def is_prime(n):
    if n<= 1 :
        print("number should be greather than 0 and 1")
        return False
    elif n%2==0:
        return False
    elif  n==2:
        return True
    for i in range(3, int(n**0.5)+1,2):
        if n%i == 0:
            return False
    return True
     
```

```python
is_prime(3)
```

```python
# Create def calculate_discount(price, discount=10)

#where the default discount is 10%.

def calculate_discount(price, discount=10):
    discount_amount = price * discount / 100
    final_price = price - discount_amount
    return final_price

```

```python
calculate_discount(100)
```

```python
# Create a function that accepts a list and returns its sum without using sum().

def sum_of_num(l):
    temp =0
    for num in l:
        temp = num+temp
    return temp
```

```python
# Create a function that accepts a string and returns the number of vowels.

def count_vowels(s):
    vowels = frozenset("aeiou")
    count = 0

    for char in s:
        if char in vowels:
            count+=1

    return count
```

```python
# Create a function that accepts a string and determines whether it is a palindrome.

def is_palindrome(s):
    reversed = []
    n = len(s)

    for i in range(n-1, -1, -1):
        reversed.append(s[i])   

    reversed_string = "".join(reversed) 

    return True if reversed_string == s else False
```

```python
is_palindrome("lal") 
```
