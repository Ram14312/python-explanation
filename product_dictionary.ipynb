```python
laptop = {"brand": "Dell", "model": "XPS 15", "price": 120000, "ram": "16GB","storage": "512GB SSD","available": True}
```

```python
# Display the product
print(laptop)
```

```python
# Print brand
print(laptop['brand'])
```

```python
# Print model
print(laptop['model'])
```

```python
# Print price
print(laptop['price'])
```

```python
# Change price
laptop['price'] = 90000
```

```python
print(laptop)
```

```python
# Add processor information
laptop['processor_info'] = 'i5'
```

```python
print(laptop)
```

```python
# Add GPU information
laptop['GPU_info'] = '4GB'
```

```python
print(laptop)
```

```python
# Change RAM to "32GB"
laptop['ram'] = "32GB"
```

```python
print(laptop)
```

```python
# Remove "available"
laptop.pop("available")
```

```python
print(laptop)
```

```python
# Print all keys
print(laptop.keys())
```

```python
# Print all values
print(laptop.values())
```

```python
#Print all items
print(laptop.items())
```
